# Legacy-to-Revamp Record Match

This notebook performs a record-level comparison between the legacy thesis final dataset (~20,699 mentions) and the revamp pipeline output (53,892 mentions). The goal is to determine whether every observation in the legacy dataset exists somewhere in the revamp output â€” confirming the revamp is a superset of the legacy.

```python
import pandas as pd
import numpy as np
from pathlib import Path

DATA_DIR = Path("../data/output")

# Load legacy dataset (comma-separated .txt file with level1_ prefix on all columns)
legacy = pd.read_csv(DATA_DIR / "data_legacy_thesis.txt", sep=",")

# Load revamp dataset
revamp = pd.read_csv(DATA_DIR / "level1.csv")

print(f"Legacy: {legacy.shape[0]:,} rows x {legacy.shape[1]} columns")
print(f"Revamp: {revamp.shape[0]:,} rows x {revamp.shape[1]} columns")
```

```python
# Column mapping table: legacy -> revamp
# Strip the level1_ prefix from legacy columns and find matches in revamp
legacy_cols_stripped = {c: c.replace("level1_", "", 1) for c in legacy.columns}
revamp_col_set = set(revamp.columns)

rows = []
for leg_col, stripped in sorted(legacy_cols_stripped.items()):
    if stripped in revamp_col_set:
        rows.append({"legacy_column": leg_col, "revamp_column": stripped, "status": "matched"})
    else:
        rows.append({"legacy_column": leg_col, "revamp_column": "\u2014", "status": "legacy only"})

# Revamp-only columns
legacy_stripped_set = set(legacy_cols_stripped.values())
for rc in sorted(revamp.columns):
    if rc not in legacy_stripped_set:
        rows.append({"legacy_column": "\u2014", "revamp_column": rc, "status": "revamp only"})

mapping_df = pd.DataFrame(rows)
print(f"Matched: {(mapping_df['status']=='matched').sum()}  |  "
      f"Legacy only: {(mapping_df['status']=='legacy only').sum()}  |  "
      f"Revamp only: {(mapping_df['status']=='revamp only').sum()}")
mapping_df
```

## Section 2: Match on (org_id, granuleId)

This is the core test. We normalize keys, create unique pairs in each dataset, and perform an outer merge to see how many legacy pairs appear in the revamp.

```python
# 2a: Normalize keys
# Legacy columns have level1_ prefix
LEGACY_ORG = "level1_org_id"
LEGACY_GRANULE = "level1_granuleId"
REVAMP_ORG = "org_id"
REVAMP_GRANULE = "granuleId"

print(f"Legacy  org column:     {LEGACY_ORG}")
print(f"Legacy  granule column: {LEGACY_GRANULE}")
print(f"Revamp  org column:     {REVAMP_ORG}")
print(f"Revamp  granule column: {REVAMP_GRANULE}")

# Cast to string and strip whitespace
legacy["_org_id"] = legacy[LEGACY_ORG].astype(str).str.strip()
legacy["_granuleId"] = legacy[LEGACY_GRANULE].astype(str).str.strip()

revamp["_org_id"] = revamp[REVAMP_ORG].astype(str).str.strip()
revamp["_granuleId"] = revamp[REVAMP_GRANULE].astype(str).str.strip()

# Normalize org_id: legacy may have float-like strings (e.g. "1453.0") -- strip trailing .0
legacy["_org_id"] = legacy["_org_id"].str.replace(r"\.0$", "", regex=True)
revamp["_org_id"] = revamp["_org_id"].str.replace(r"\.0$", "", regex=True)

print(f"\nLegacy org_id sample:    {legacy['_org_id'].iloc[:3].tolist()}")
print(f"Revamp org_id sample:    {revamp['_org_id'].iloc[:3].tolist()}")
print(f"Legacy granuleId sample: {legacy['_granuleId'].iloc[:3].tolist()}")
print(f"Revamp granuleId sample: {revamp['_granuleId'].iloc[:3].tolist()}")
```

```python
# 2b: Create unique (org_id, granuleId) pairs
legacy_pairs = legacy[["_org_id", "_granuleId"]].drop_duplicates().rename(
    columns={"_org_id": "org_id", "_granuleId": "granuleId"}
)
revamp_pairs = revamp[["_org_id", "_granuleId"]].drop_duplicates().rename(
    columns={"_org_id": "org_id", "_granuleId": "granuleId"}
)

print(f"Legacy unique (org_id, granuleId) pairs: {len(legacy_pairs):,}")
print(f"Revamp unique (org_id, granuleId) pairs: {len(revamp_pairs):,}")
```

```python
# 2c: Perform the match
merged = legacy_pairs.merge(revamp_pairs, on=["org_id", "granuleId"], how="outer", indicator=True)

counts = merged["_merge"].value_counts()
both = counts.get("both", 0)
left_only = counts.get("left_only", 0)
right_only = counts.get("right_only", 0)
total_legacy = len(legacy_pairs)

print("=" * 60)
print("RECORD MATCH RESULTS")
print("=" * 60)
print(f"Total unique (org_id, granuleId) pairs in legacy:  {total_legacy:,}")
print(f"Total unique (org_id, granuleId) pairs in revamp:  {len(revamp_pairs):,}")
print(f"")
print(f"Pairs in BOTH:                {both:,}  ({both/total_legacy*100:.1f}% of legacy)")
print(f"Pairs in LEGACY ONLY:         {left_only:,}  ({left_only/total_legacy*100:.1f}% of legacy)")
print(f"Pairs in REVAMP ONLY:         {right_only:,}")
print(f"")
if both / total_legacy >= 0.95:
    print(f">>> SUPERSET CONFIRMED: {both/total_legacy*100:.1f}% of legacy pairs found in revamp.")
else:
    print(f">>> WARNING: Only {both/total_legacy*100:.1f}% of legacy pairs found in revamp.")
```

## Section 3: Analyze Unmatched Legacy Rows

For legacy (org_id, granuleId) pairs NOT found in the revamp â€” how many, which orgs, which congresses, and diagnostic checks.

```python
# Get the missing legacy pairs
missing_pairs = merged[merged["_merge"] == "left_only"][["org_id", "granuleId"]].copy()
print(f"Missing legacy pairs: {len(missing_pairs):,} ({len(missing_pairs)/total_legacy*100:.1f}% of legacy)\n")

# Enrich missing pairs with legacy metadata for analysis
missing_enriched = missing_pairs.merge(
    legacy[["_org_id", "_granuleId", "level1_org_name", "level1_congress"]].drop_duplicates(
        subset=["_org_id", "_granuleId"]
    ).rename(columns={"_org_id": "org_id", "_granuleId": "granuleId"}),
    on=["org_id", "granuleId"],
    how="left"
)

# Top 15 orgs by missing count
print("Top 15 organizations by count of missing pairs:")
top_orgs = (
    missing_enriched.groupby(["org_id", "level1_org_name"])
    .size()
    .reset_index(name="missing_count")
    .sort_values("missing_count", ascending=False)
    .head(15)
)
print(top_orgs.to_string(index=False))
```

```python
# Distribution by congress
print("Missing pairs by congress:")
print(missing_enriched["level1_congress"].value_counts().sort_index())

# Concentration: how many unique granules vs unique orgs?
print(f"\nMissing pairs span {missing_enriched['granuleId'].nunique():,} unique granules "
      f"and {missing_enriched['org_id'].nunique():,} unique orgs")
```

```python
# 10 example missing pairs with all available legacy columns
print("10 example missing pairs:")
examples = missing_pairs.head(10).merge(
    legacy.rename(columns={"_org_id": "org_id", "_granuleId": "granuleId"}),
    on=["org_id", "granuleId"],
    how="left"
).drop_duplicates(subset=["org_id", "granuleId"])

# Show key columns only for readability
key_cols = ["org_id", "granuleId", "level1_org_name", "level1_congress",
            "level1_acronym", "level1_variation"]
avail = [c for c in key_cols if c in examples.columns]
examples[avail]
```

```python
# Diagnostic checks
revamp_org_set = set(revamp["_org_id"].unique())
revamp_granule_set = set(revamp["_granuleId"].unique())

missing_org_in_revamp = missing_pairs["org_id"].isin(revamp_org_set)
missing_granule_in_revamp = missing_pairs["granuleId"].isin(revamp_granule_set)

org_exists = missing_org_in_revamp.sum()
granule_exists = missing_granule_in_revamp.sum()
both_exist = (missing_org_in_revamp & missing_granule_in_revamp).sum()
neither_exists = (~missing_org_in_revamp & ~missing_granule_in_revamp).sum()
org_only = (missing_org_in_revamp & ~missing_granule_in_revamp).sum()
granule_only = (~missing_org_in_revamp & missing_granule_in_revamp).sum()

print("DIAGNOSTIC: For missing legacy pairs, do the individual keys exist in revamp?\n")
print(f"  Org exists in revamp (different granule):     {org_exists:,}")
print(f"  Granule exists in revamp (different org):      {granule_exists:,}")
print(f"  BOTH org AND granule exist (just not paired):  {both_exist:,}")
print(f"  NEITHER org NOR granule exists in revamp:       {neither_exists:,}")
print(f"  Only org exists (granule missing entirely):     {org_only:,}")
print(f"  Only granule exists (org missing entirely):     {granule_only:,}")
print(f"\nSummary: {org_exists:,} missing pairs where org exists in revamp, "
      f"{granule_exists:,} where granule exists, {neither_exists:,} where neither exists.")
```

## Section 4: Profile Revamp-Only Rows

For (org_id, granuleId) pairs in revamp but NOT in legacy â€” where do the extra mentions come from?

```python
# Revamp-only pairs
revamp_only_pairs = merged[merged["_merge"] == "right_only"][["org_id", "granuleId"]].copy()
print(f"Revamp-only (org_id, granuleId) pairs: {len(revamp_only_pairs):,}\n")

# Enrich with revamp metadata
revamp_only_enriched = revamp_only_pairs.merge(
    revamp[["_org_id", "_granuleId", "org_name", "congress", "is_acronym"]].drop_duplicates(
        subset=["_org_id", "_granuleId"]
    ).rename(columns={"_org_id": "org_id", "_granuleId": "granuleId"}),
    on=["org_id", "granuleId"],
    how="left"
)

# Top 20 orgs
print("Top 20 organizations in revamp-only pairs:")
top_revamp_orgs = (
    revamp_only_enriched.groupby(["org_id", "org_name"])
    .size()
    .reset_index(name="count")
    .sort_values("count", ascending=False)
    .head(20)
)
print(top_revamp_orgs.to_string(index=False))
```

```python
# Acronym vs full-name matches in revamp-only
if "is_acronym" in revamp_only_enriched.columns:
    print("Revamp-only pairs by is_acronym:")
    acr_counts = revamp_only_enriched["is_acronym"].value_counts()
    for val, cnt in acr_counts.items():
        print(f"  is_acronym={val}: {cnt:,} ({cnt/len(revamp_only_enriched)*100:.1f}%)")
else:
    print("No is_acronym column available in revamp-only enriched data.")

print(f"\nRevamp-only pairs by congress:")
print(revamp_only_enriched["congress"].value_counts().sort_index())
```

## Section 5: Finer-Grained Match (paragraph text comparison)

```python
# Check for paragraph/text columns in both datasets
legacy_text_cols = [c for c in legacy.columns if any(kw in c.lower() for kw in ["paragraph", "sentence", "p1_original", "text"])]
revamp_text_cols = [c for c in revamp.columns if any(kw in c.lower() for kw in ["paragraph", "sentence", "text"])]

print(f"Legacy text-like columns: {legacy_text_cols}")
print(f"Revamp text-like columns: {revamp_text_cols}")

# Both have paragraph columns
LEGACY_TEXT = "level1_p1_original"
REVAMP_TEXT = "paragraph"

if LEGACY_TEXT in legacy.columns and REVAMP_TEXT in revamp.columns:
    print(f"\nUsing legacy=\"{LEGACY_TEXT}\" and revamp=\"{REVAMP_TEXT}\" for text comparison.")

    # For matched pairs, compare paragraph text using first 80 chars
    matched_pairs = merged[merged["_merge"] == "both"][["org_id", "granuleId"]]

    # Get first legacy text per pair
    leg_text = legacy[["_org_id", "_granuleId", LEGACY_TEXT]].rename(
        columns={"_org_id": "org_id", "_granuleId": "granuleId", LEGACY_TEXT: "legacy_text"}
    ).drop_duplicates(subset=["org_id", "granuleId"])
    leg_text["legacy_text_80"] = leg_text["legacy_text"].astype(str).str[:80].str.strip().str.lower()

    # Get first revamp text per pair
    rev_text = revamp[["_org_id", "_granuleId", REVAMP_TEXT]].rename(
        columns={"_org_id": "org_id", "_granuleId": "granuleId", REVAMP_TEXT: "revamp_text"}
    ).drop_duplicates(subset=["org_id", "granuleId"])
    rev_text["revamp_text_80"] = rev_text["revamp_text"].astype(str).str[:80].str.strip().str.lower()

    text_compare = matched_pairs.merge(leg_text[["org_id", "granuleId", "legacy_text_80", "legacy_text"]],
                                        on=["org_id", "granuleId"], how="left")
    text_compare = text_compare.merge(rev_text[["org_id", "granuleId", "revamp_text_80", "revamp_text"]],
                                       on=["org_id", "granuleId"], how="left")

    text_compare["text_match"] = text_compare["legacy_text_80"] == text_compare["revamp_text_80"]
    match_rate = text_compare["text_match"].mean()

    print(f"\nOf {len(text_compare):,} matched (org_id, granuleId) pairs:")
    print(f"  Text also matches (first 80 chars): {text_compare['text_match'].sum():,} ({match_rate*100:.1f}%)")
    print(f"  Text differs:                       {(~text_compare['text_match']).sum():,} ({(1-match_rate)*100:.1f}%)")

    # Show 5 examples where text differs
    mismatches = text_compare[~text_compare["text_match"]].head(5)
    if len(mismatches) > 0:
        print(f"\n5 examples where (org_id, granuleId) matches but paragraph text differs:")
        for _, row in mismatches.iterrows():
            print(f"\n  org_id={row['org_id']}, granuleId={row['granuleId']}")
            print(f"  Legacy:  {str(row['legacy_text'])[:120]}...")
            print(f"  Revamp:  {str(row['revamp_text'])[:120]}...")
else:
    print("\nText-level comparison not possible -- required columns not found in both datasets.")
```

## Section 7: Corrected Analysis (Excluding Zero-Mention Orgs)

The legacy dataset includes organizations with zero congressional mentions by design. The thesis studied what predicts prominence, requiring the full population of WRS organizations as the denominator. Organizations never mentioned in the Congressional Record have a prominence value of zero. These rows result from a left join of the full WRS dictionary onto mention-level data, not from a data error.

The original analysis in Sections 2-6 above reported a 58.6% match rate because it included these zero-mention rows as legacy pairs with `granuleId = NaN`. Since those rows have no document match by construction, they inflate the denominator. This section corrects the analysis by filtering to mention-bearing rows only.

```python
# Filter legacy to rows with actual mentions (non-null granuleId)
legacy_raw_count = len(legacy)
has_granule = legacy['level1_granuleId'].notna() & (legacy['level1_granuleId'].astype(str) != 'nan')
legacy_mentions = legacy[has_granule].copy()

print('Legacy dataset filtering:')
print(f'  Rows before filtering:     {legacy_raw_count:,}')
print(f'  Rows after filtering:      {len(legacy_mentions):,}')
print(f'  Zero-mention rows removed: {legacy_raw_count - len(legacy_mentions):,}')
print(f'  Unique org_ids (all rows):       {legacy["_org_id"].nunique():,}')
print(f'  Unique org_ids (mentions only):  {legacy_mentions["_org_id"].nunique():,}')
print(f'  Zero-mention orgs removed:       {legacy["_org_id"].nunique() - legacy_mentions["_org_id"].nunique():,}')
```

```python
# Re-run (org_id, granuleId) pair match on filtered legacy data
legacy_pairs_corr = legacy_mentions[['_org_id', '_granuleId']].drop_duplicates().rename(
    columns={'_org_id': 'org_id', '_granuleId': 'granuleId'}
)
# revamp_pairs already defined in Section 2

merged_corr = legacy_pairs_corr.merge(revamp_pairs, on=['org_id', 'granuleId'], how='outer', indicator=True)
counts_corr = merged_corr['_merge'].value_counts()
both_corr = counts_corr.get('both', 0)
left_only_corr = counts_corr.get('left_only', 0)
right_only_corr = counts_corr.get('right_only', 0)
total_legacy_corr = len(legacy_pairs_corr)

print('=' * 60)
print('CORRECTED RECORD MATCH RESULTS (mentions only)')
print('=' * 60)
print(f'Legacy unique (org_id, granuleId) pairs:  {total_legacy_corr:,}')
print(f'Revamp unique (org_id, granuleId) pairs:  {len(revamp_pairs):,}')
print()
print(f'Pairs in BOTH:           {both_corr:,}  ({both_corr/total_legacy_corr*100:.1f}% of legacy)')
print(f'Pairs in LEGACY ONLY:    {left_only_corr:,}  ({left_only_corr/total_legacy_corr*100:.1f}% of legacy)')
print(f'Pairs in REVAMP ONLY:    {right_only_corr:,}')
print()
if both_corr / total_legacy_corr >= 0.95:
    print(f'>>> SUPERSET CONFIRMED: {both_corr/total_legacy_corr*100:.1f}% of legacy mention pairs found in revamp.')
else:
    print(f'>>> Match rate: {both_corr/total_legacy_corr*100:.1f}% of legacy mention pairs found in revamp.')
```

```python
# Investigate the truly missing orgs (have mentions in legacy, absent from revamp)
missing_corr = merged_corr[merged_corr['_merge'] == 'left_only'][['org_id', 'granuleId']].copy()

# Which orgs are in legacy mentions but not in revamp at all?
legacy_mention_orgs = set(legacy_mentions['_org_id'].unique())
revamp_orgs = set(revamp['_org_id'].unique())
missing_orgs = legacy_mention_orgs - revamp_orgs

print(f'Legacy orgs with mentions: {len(legacy_mention_orgs):,}')
print(f'Revamp orgs:               {len(revamp_orgs):,}')
print(f'In both:                   {len(legacy_mention_orgs & revamp_orgs):,}')
print(f'Legacy-only (truly missing): {len(missing_orgs):,}')
print(f'Revamp-only (new):           {len(revamp_orgs - legacy_mention_orgs):,}')

# Count mentions per missing org
missing_org_mentions = (
    legacy_mentions[legacy_mentions['_org_id'].isin(missing_orgs)]
    .groupby(['_org_id', 'level1_org_name'])
    .size()
    .reset_index(name='legacy_mentions')
    .sort_values('legacy_mentions', ascending=False)
)
print(f'\nTotal mention rows from missing orgs: {missing_org_mentions["legacy_mentions"].sum():,}')
print(f'({missing_org_mentions["legacy_mentions"].sum()/len(legacy_mentions)*100:.1f}% of legacy mentions)\n')
print('All missing orgs with mention counts:')
print(missing_org_mentions.head(40).to_string(index=False))
```

```python
# Fuzzy-match missing org names against revamp org names
import difflib

revamp_org_names = revamp[['_org_id', 'org_name']].drop_duplicates(subset=['_org_id'])
revamp_name_list = revamp_org_names['org_name'].dropna().unique().tolist()

fuzzy_results = []
for _, row in missing_org_mentions.iterrows():
    legacy_name = str(row['level1_org_name']).strip()
    if not legacy_name or legacy_name == 'nan':
        fuzzy_results.append({
            'org_id': row['_org_id'],
            'legacy_name': legacy_name,
            'legacy_mentions': row['legacy_mentions'],
            'closest_revamp_name': '(no name)',
            'similarity': 0.0,
            'likely_cause': 'no_name',
        })
        continue
    matches = difflib.get_close_matches(legacy_name, revamp_name_list, n=1, cutoff=0.0)
    if matches:
        best = matches[0]
        sim = difflib.SequenceMatcher(None, legacy_name.lower(), best.lower()).ratio()
    else:
        best = '(no match)'
        sim = 0.0
    if sim > 0.6:
        cause = 'name_change'
    elif sim > 0.3:
        cause = 'partial_match'
    else:
        cause = 'genuinely_absent'
    fuzzy_results.append({
        'org_id': row['_org_id'],
        'legacy_name': legacy_name,
        'legacy_mentions': row['legacy_mentions'],
        'closest_revamp_name': best,
        'similarity': round(sim, 3),
        'likely_cause': cause,
    })

fuzzy_df = pd.DataFrame(fuzzy_results)

# Summary
cause_counts = fuzzy_df['likely_cause'].value_counts()
print('MISSING ORG DIAGNOSIS')
print('=' * 60)
for cause, cnt in cause_counts.items():
    mentions = fuzzy_df[fuzzy_df['likely_cause'] == cause]['legacy_mentions'].sum()
    print(f'  {cause:<20} {cnt:>4} orgs  ({mentions:>4} mentions)')

print(f'\nTop 30 missing orgs with fuzzy matches:')
display_cols = ['org_id', 'legacy_name', 'legacy_mentions', 'closest_revamp_name', 'similarity', 'likely_cause']
print(fuzzy_df[display_cols].head(30).to_string(index=False))
```
